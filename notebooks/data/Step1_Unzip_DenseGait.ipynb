**Data prep.** Unzips the password-protected `DenseGait.zip` inside the Drive project folder.

The archive password is issued by the DenseGait authors on request and has been replaced with a placeholder. Never commit the real password.

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
import os
os.chdir('/content/drive/MyDrive/Gait Project')
```

```python
import zipfile
import os

zip_file_path = 'DenseGait.zip'
password = b'<DENSEGAIT_ZIP_PASSWORD>' # Password needs to be in bytes

if os.path.exists(zip_file_path):
    try:
        with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
            zip_ref.extractall(pwd=password)
        print(f"Successfully unzipped {zip_file_path}")
    except zipfile.BadZipFile:
        print(f"Error: {zip_file_path} is not a valid zip file.")
    except RuntimeError as e:
        print(f"Error unzipping {zip_file_path}: {e}")
else:
    print(f"Error: {zip_file_path} not found in the current directory.")
```
