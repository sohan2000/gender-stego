**Data prep.** Loads and plots one DenseGait sample. Each file is `(T, 18, 3)` = frames x OpenPose-18 joints x (x, y, confidence).

```python
import numpy as np

# Replace 'your_file.npy' with the actual path to your .npy file
file_path = '/content/10.npy'

try:
    data = np.load(file_path)
    print(f"Successfully loaded data from {file_path}:")
    print("Data shape:", data.shape)
    print("Data type:", data.dtype)
    print("\nFirst few elements (or full array if small):\n", data)
except FileNotFoundError:
    print(f"Error: The file '{file_path}' was not found. Please ensure the path is correct.")
except Exception as e:
    print(f"An error occurred while loading the file: {e}")
```

### Visualization 1: Single Frame Pose

Let's visualize the pose of a person in a specific frame. We'll assume the first two components (index 0 and 1) are the x and y coordinates, respectively.

```python
import matplotlib.pyplot as plt

# Choose a frame to visualize (e.g., the first frame)
frame_idx = 0

# Extract x and y coordinates for all keypoints in this frame
x_coords = data[frame_idx, :, 0]
y_coords = data[frame_idx, :, 1]

plt.figure(figsize=(8, 8))
plt.scatter(x_coords, y_coords, s=50, marker='o', alpha=0.8)

# Optionally, add labels for keypoints if you have a mapping
# For now, let's just show the points
for i, (x, y) in enumerate(zip(x_coords, y_coords)):
    plt.text(x, y, str(i), fontsize=9, ha='right')

plt.title(f'Pose Estimation for Frame {frame_idx}')
plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box') # Keep aspect ratio equal
plt.show()
```

### Visualization 2: Trajectory of a Single Keypoint Over All Frames

Now, let's see how a specific keypoint moves throughout the video. We'll track the first keypoint (index 0) over all frames.

```python
import matplotlib.pyplot as plt

# Choose a keypoint to track (e.g., the first keypoint, index 0)
keypoint_idx = 0

# Extract x and y coordinates for this keypoint across all frames
x_trajectory = data[:, keypoint_idx, 0]
y_trajectory = data[:, keypoint_idx, 1]

plt.figure(figsize=(10, 6))
plt.plot(x_trajectory, y_trajectory, marker='.', linestyle='-', alpha=0.7)
plt.title(f'Trajectory of Keypoint {keypoint_idx} Over All Frames')
plt.xlabel('X Coordinate')
plt.ylabel('Y Coordinate')
plt.grid(True)
plt.show()
```
