**Stage 1.** Prints the `GaitSTGCN` architecture and a `torchinfo` summary (2,035,905 params, ~852.67 M mult-adds for input `[1, 3, 64, 18, 1]`).

Needs `stgcn_new_optimized.py` (in this folder; the model code from the 30k run of `stgcn_final_code.ipynb`) and `model_epoch_20.pth`, the epoch-20 checkpoint of that run (72.65% val accuracy).

```python
import torch
from stgcn_new_optimized import GaitSTGCN  # Corrected model import

# Instantiate your model as it was trained
model = GaitSTGCN(in_channels=3, num_class=1, num_joints=18)

# Load state dict from checkpoint
checkpoint = torch.load('model_epoch_20.pth', map_location='cpu', weights_only=False)  # epoch 20 of the 30k run (72.65%)
model.load_state_dict(checkpoint['model_state_dict'])
model.eval()  # Set in eval mode for inference
```
