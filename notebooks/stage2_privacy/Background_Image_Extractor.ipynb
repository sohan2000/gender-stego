**Stage 2: background model.** Per-pixel **median** (BGR) of up to 100 frames sampled evenly across the video, saved as `<vid>-BG_frame.jpg`.

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# ── Config ────────────────────────────────────────────────────────────────────
INPUT_VIDEO  = 'vid10.avi'
OUTPUT_IMAGE = 'vid10-BG_frame.jpg'

# Max frames to sample (median needs all in RAM — reduce if video is very long)
MAX_FRAMES   = 100

# ─────────────────────────────────────────────────────────────────────────────

cap = cv2.VideoCapture(INPUT_VIDEO)
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
print(f"Video has {total_frames} frames. Sampling up to {MAX_FRAMES}.")

# Sample evenly spaced frame indices across the full video
sample_indices = np.linspace(0, total_frames - 1, min(MAX_FRAMES, total_frames), dtype=int)

frames = []
for idx in sample_indices:
    cap.set(cv2.CAP_PROP_POS_FRAMES, idx)
    ret, frame = cap.read()
    if ret:
        frames.append(frame)

cap.release()
print(f"Collected {len(frames)} frames.")

# Stack into array and take median across frame axis
frames_array = np.array(frames, dtype=np.uint8)
background   = np.median(frames_array, axis=0).astype(np.uint8)

cv2.imwrite(OUTPUT_IMAGE, background)
print(f"✓ Background saved → {OUTPUT_IMAGE}")

# Optional: show it
cv2_imshow(background)
```
