**Stage 3.** Plots per-frame PSNR / MSE / similarity from a `Results*.xlsx` metrics sheet.

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# ==== LOAD DATA FROM XLSX ====
file_path = "vid2-ResultsQP27.xlsx"
sheet_name = "QP27B30"  # Update if needed

df = pd.read_excel(file_path, sheet_name=sheet_name)

# Expecting columns:
# "Frame","MSE (Stego)","PSNR (Stego)","Similarity (Stego)",
# "MSE (Extracted)","PSNR (Extracted)","Similarity (Extracted)"

frame = df["Frame"]

# ==== Create directory for plots ====
output_dir = "graphs"
if not os.path.exists(output_dir):
    os.makedirs(output_dir)

# ==== PLOTS: Scores vs Frames ====

# PSNR Plot
plt.figure(figsize=(12,8))
plt.plot(frame, df["PSNR (Stego)"], marker='o', label='PSNR (Stego)', linewidth=2)
plt.plot(frame, df["PSNR (Extracted)"], marker='x', label='PSNR (Extracted)', linewidth=2)
plt.xlabel('Frame')
plt.ylabel('PSNR')
plt.title('PSNR vs Frame')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "PSNR_vs_Frame.png"))
plt.show()

# MSE Plot
plt.figure(figsize=(12,8))
plt.plot(frame, df["MSE (Stego)"], marker='o', label='MSE (Stego)', linewidth=2)
plt.plot(frame, df["MSE (Extracted)"], marker='x', label='MSE (Extracted)', linewidth=2)
plt.xlabel('Frame')
plt.ylabel('MSE')
plt.title('MSE vs Frame')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "MSE_vs_Frame.png"))
plt.show()

# Similarity Plot
plt.figure(figsize=(12,8))
plt.plot(frame, df["Similarity (Stego)"], marker='o', label='Similarity (Stego)', linewidth=2)
plt.plot(frame, df["Similarity (Extracted)"], marker='x', label='Similarity (Extracted)', linewidth=2)
plt.xlabel('Frame')
plt.ylabel('Similarity (%)')
plt.title('Similarity vs Frame')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig(os.path.join(output_dir, "Similarity_vs_Frame.png"))
plt.show()
```
